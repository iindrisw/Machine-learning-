1. Apa perbedaan supervised, unsupervised, dan reinforcement learning?


supervised itu data yang mempunyai label misal mahasiswa dengan label jenis kelamin
unsupervised itu data yang tidak mempunyai label
reinformance itu itu belajar lewat sistem Trial & Error (coba-coba) menggunakan Hadiah (Reward) dan Hukuman (Penalty).

2. Mengapa pembagian data latih dan uji penting?


pembagian data latih dengan data uji penting karena untuk mengukur tingkat keberhasilannya sudah berapa persen. Juga agar model paham tidak hanya menghafal pola lama jadi kita kasih data latih dan kita uji dengan data uji jika memang model betul memahami maka akan tahu pola jawabannya, jika hanya menghafal maka model tidak akan tahu apa itu data uji

3. Mengapa alur kerja ML disebut iteratif?


disebut iteratif karena alur kerjanya berbentuk lingkaran coba-evaluasi-perbaiki (feedback loop) di mana model terus bolak-balik antar-tahap sampai hasilnya layak digunakan di dunia nyata
